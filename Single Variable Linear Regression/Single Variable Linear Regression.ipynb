```python
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
from sklearn import metrics
import pandas as pd
```

```python
x_train = pd.read_csv('C:/Users/Muhammad Darab/OneDrive/Desktop/Online Course Files/Machine Learning/Data Sets/SingleVariableLinearRegression/SinVarLinReg.csv', usecols = ['Temp'])
y_train = pd.read_csv('C:/Users/Muhammad Darab/OneDrive/Desktop/Online Course Files/Machine Learning/Data Sets/SingleVariableLinearRegression/SinVarLinReg.csv', usecols = ['Salinity'])
x_train = x_train.to_numpy()
y_train = y_train.to_numpy()
x_train = x_train[:2500]
y_train = y_train[:2500]
print(x_train.shape, y_train.shape)
plt.scatter(x_train, y_train, marker='x', c='r') 

plt.title("Salinity vs. Temperature")

plt.ylabel('Salinity')

plt.xlabel('Temperature')
plt.show()
```

```python
def costFunc(w1, b, lambda_):
    J = 0
    y_pred = []
    m, n = x_train.shape
    regu = (w1 ** 2)
    regu = (lambda_/(2 * m)) * regu
    for i in range(m):
        f_x = w1 * x_train[i] + b
        y_pred.append(f_x)
        J += (f_x - y_train[i])**2
    J = ((1 / (2*m)) * J) + regu
    return J, y_pred
```

```python
def computeGradient(w1, b, lambda_):
    m, n = x_train.shape
    dJ_dw_1 = 0
    dJ_db = 0
    for i in range(m):
        f_x = w1 * x_train[i] + b
        dJ_dw_1 += (f_x - y_train[i]) * x_train[i]
        dJ_db += (f_x - y_train[i])
    dJ_dw_1 = ((1 / m) * dJ_dw_1) + ((lambda_ / m) * w1)
    dJ_db = (1 / m) * dJ_db
    return dJ_dw_1, dJ_db
```

```python
def gradientDes(alpha, iterations, w1, b, lambda_):
    for i in range(iterations):
        dJ_dw_1, dJ_db = computeGradient(w1, b, lambda_)
        w1 = w1 - alpha * dJ_dw_1
        b = b - alpha * dJ_db
        if i % 100 == 0:
            J, y_pred = costFunc(w1, b, lambda_)
            print("At iteration", i, " Cost is ", J)
    return w1, b
```

```python
def accuracy(y_pred, y_train):
    accuracy = []
    for i in range(len(y_pred)):
        accuracy.append(abs((y_pred[i] - y_train[i]) / y_train[i]))
    accu = np.mean(accuracy)
    return accu
```

```python
w1, b = gradientDes(0.0202, 3000, 1, 3, 0.1)
print("Optimal values of w and b are ", w1, b)
```

```python
J, y_pred = costFunc(w1, b, 0.1)
print("The accuracy of the Linear Regression Salinity vs Temperature Model is", round(100 - accuracy(y_pred, y_train) * 100, 2), "%")
```
