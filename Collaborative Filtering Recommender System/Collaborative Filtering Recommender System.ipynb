```python
def createVariables(Y, nullVal, n_u, n_j):
    s = (n_u, n_j)
    R = np.array(np.zeros(s))
    for i in range(n_u):
        for j in range(n_j):
            if Y[i][j] != nullVal:
                R[i][j] = 1
            else:
                R[i][j] = 0
    return R
```

```python
def costFunc1(X, W, b, Y, R, lambda_):
    j = (tf.linalg.matmul(X, tf.transpose(W)) + b - Y)*R
    J = 0.5 * tf.reduce_sum(j**2) + (lambda_/2) * (tf.reduce_sum(X**2) + tf.reduce_sum(W**2))
    return J
```

```python
def gradients(R, Y, W, X, b, lambda_, n_u, n_j):
    dJ_dw = 0
    dJ_dx = 0
    dJ_db = 0
    J = 0
    regu1 = 0
    regu2 = 0
    n_1, m_1 = X.shape
    n_2, m_2 = W.shape
    for j in range(n_u):
        for i in range(m_2):
            regu1 += (W[j][i])
    for j in range(n_j):
        for i in range(m_1):
            regu2 += (X[j][i])
    regu1 = (lambda_) * regu1
    regu2 = (lambda_) * regu2
    for i in range(n_j):
        for j in range(n_u):
            f_x = np.dot(W[j], X[i]) + b[0][j]
            dJ_dw += R[j][i] * (f_x - Y[j][i]) * X[i]
            dJ_dx += R[j][i] * (f_x - Y[j][i]) * W[j]
            dJ_db += R[j][i] * (f_x - Y[j][i])
    dJ_dw += regu1
    dJ_dx += regu2
    return dJ_dw, dJ_dx, dJ_db
```

```python
def gradientDescent(iterations, R, Y, W, X, b, lambda_, alpha, n_u, n_j):
    for i in range(iterations):
        dJ_dw, dJ_dx, dJ_db = gradients(R, Y, W, X, b, lambda_, n_u, n_j)
        W = W - alpha * dJ_dw
        X = X - alpha * dJ_dx
        b = b - alpha * dJ_db
        J = costFunc(R, X, W, b, Y, lambda_, n_u, n_j)
        print("Cost at iteration", i, "is", J)
    return W, X, b 
```

```python
def contentBasedFilteringRecommender(Y, nullVal, iterations, alpha):
    n_u, n_j = Y.shape
    R = createVariables(Y, nullVal, n_u, n_j)
    num_features = 150
    W = np.random.rand(n_u, num_features)
    X = np.random.rand(n_j, num_features)
    b = np.random.rand(1, n_u)
    W, X, b = gradientDescent(iterations, R, Y, W, X, b, 0.5, alpha, n_u, n_j)
```
